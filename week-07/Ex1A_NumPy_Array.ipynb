## Looping Through Arrays

```python
arrA = [
    [11, 12, 13],
    [14, 15, 16]
]
```

```python
print(arrA)
```

```python
for x in arrA:
    print(x*2)
```

```python
for y in x:
    print(y*2)
```

The first loop prints the array twice as informed by the (x*2). The second loop behaves differently by multiplying the y values in the array by two as opposed to printing it twice.

## Joining and Stacking Array

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.concatenate((joined_array_1, joined_array_1), axis=0)
print(stacked_array_1)
```

```python
stacked_array_2 = np.concatenate((joined_array_1, joined_array_2), axis=0)
print(stacked_array_2)

#ValueError
# The array dimensions must match exactly in order to be stacked. the arrays do not match along dimension one.
```

```python
VarH = np.hstack((arrA, arrA))
VarV = np.vstack((arrA, arrA))
VarD = np.dstack((arrA, arrA))
```

```python
print(f" hstack \n {VarH}") # Horizontal stacking of arrays
print("\n ~~~~~~~~~~~")
print(f" vstack \n {VarV}") # Vertical stacking of arrays
print("\n ~~~~~~~~~~~")
print(f" dstack \n {VarD}") # 3d stacking of arrays
```

## Sorting and Filtering Arrays

```python
# 2d array arrB
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

# Sorting arrB
arrB_sorted = np.sort(arrB)

# Print statement
print(arrB_sorted)
```

```python
# Values less than 7
z = np.where(arrB < 7)
# Filter
arrC = arrB[z]
# Print
print(arrC)
```

```python
# Values Greater than or equal to 7
d = np.where(arrB >= 7)
# Filter
arrD = arrB[d]
# Print
print(arrD)
```
