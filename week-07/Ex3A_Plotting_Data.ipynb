# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
# plt.show() Not necessary to call plot in jupyter
```

```python
plt.plot(x, y, marker='o')
```

```python
# Labeled Axis
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

## Styling Plots

```python
# The Red One
plt.plot(x, y, marker='o', linestyle='--', color='red')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# The Green One
plt.plot(x, y, marker='s', markersize='10', linestyle='--', color='green')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# Custom 1
plt.plot(x, y, marker='*', markersize='20', linestyle=':', color='yellow')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# Custom 2
plt.plot(x, y, marker='d', markersize='15', linestyle='-', color='purple')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

# Exercise 3.B More Plotting Practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
# SALES BY PRODUCT, SkyBlue
plt.bar(products, sales, width=0.5, label='Sales by Product', color='skyblue')
plt.xlabel('Products')
plt.ylabel('Sales')
```

```python
# Scatter Plot
x = np.random.rand(50)
y = np.random.rand(50)
s = np.random.rand(50) * 100
#~~~~~~~~~~~~~~~~~~~~#
plt.scatter(x, y, s=s, c=y, cmap='viridis')
plt.title('Random Chart')
plt.xlabel('Random X')
plt.ylabel('Random Y')
```

## Subplots

```python
# LINE CHART SINE WAVE
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.title('RANDOM SINE')
#~~~~~~~~~~~~~~~~~~~~~~~~~~#
plt.plot(x, y, color='red')
# RANDOM HISTOGRAM
plt.hist(np.random.randn(1000), bins=10, color='steelblue', edgecolor = 'black')
plt.title('RANDOM HISTO')
plt.xlabel('RANDY X')
plt.ylabel('RANDY Y')
```
