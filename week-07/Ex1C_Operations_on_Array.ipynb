## Permutations

```python
# Current value of r_4
print(r_4)
# Shuffled
np.random.shuffle(r_4)
print(f" Shuffled: \n {r_4}")
```

The order of the rows are shuffled while the values inside remain the same.

```python
# New shuffs
new_shuffle = np.random.permutation(r_4)
original = r_4.copy()
# Print Statements
print(f"NEW SHUFFLE \n {new_shuffle}")
print(f"ORIGINAL \n {original}")
```

Permutation is useful when you want to keep the original array intact. Shuffling your array will shuffle the output as well as the original array.

## Teamwork

```python
# Doors Knocked On
ali_doors = [5, 4, 6, 4, 2]
ben_doors = [0, 5, 5, 3, 4] # missed mondee
carla_doors = [3, 6, 0, 3, 4] # missed wednesdee

# SUM OF KNOCKS
all_knocks = np.sum(ali_doors + ben_doors + carla_doors)
print(f"ALL KNOCKS = {all_knocks}")

# Totals by knockers
ali_total = sum(ali_doors)
ben_total = sum(ben_doors)
carla_total = sum(carla_doors)

# Totals by group
group_total = ali_total + ben_total + carla_total

# Print Statements
print(f"Ali knocked on {ali_total} doors.")
print(f"Ben knocked on {ben_total} doors.")
print(f"Carla knocked on {carla_total} doors.")

# daily totals
daily_totals = []
for i in range(5):
    total = ali_doors[i] + ben_doors[i] + carla_doors[i]
    daily_totals.append(total)

# Rolling totals
rolling_totals = []
running_sum = 0

for day in daily_totals:
    running_sum += day
    rolling_totals.append(running_sum)

# Estimated votes
estimated_votes = [total * 0.5 for total in rolling_totals]

print("Rolling estimated votes by day:", estimated_votes)
```
