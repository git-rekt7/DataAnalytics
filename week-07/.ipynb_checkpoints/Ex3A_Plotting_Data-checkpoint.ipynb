# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
# plt.show() Not necessary to call plot in jupyter
```

```python
plt.plot(x, y, marker='o')
```

```python
# Labeled Axis
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

## Styling Plots

```python
# The Red One
plt.plot(x, y, marker='o', linestyle='--', color='red')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# The Green One
plt.plot(x, y, marker='s', markersize='10', linestyle='--', color='green')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# Custom 1
plt.plot(x, y, marker='*', markersize='20', linestyle=':', color='yellow')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# Custom 2
plt.plot(x, y, marker='d', markersize='15', linestyle='-', color='purple')
plt.title('Example Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```
