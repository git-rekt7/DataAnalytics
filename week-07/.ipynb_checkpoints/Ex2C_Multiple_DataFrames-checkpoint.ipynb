# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
df_sales = pd.DataFrame({'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
'West', 'North'],
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]})
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price'] # REVENUE
```

```python
print(df_sales)
```

```python
total_units_sold = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()

print(f"Total Units Sold: {total_units_sold}")
print(f"Total Revenue: {total_revenue}")
```

```python
average_units_sold = df_sales['Units_Sold'].mean().round(2)
average_revenue = df_sales['Revenue'].mean().round(2)

print(average_units_sold)
print(average_revenue)
```

## Grouping Data

```python
# Summary of each product
product_summary = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
print(product_summary)
```

```python
# Sales Averages
sales_averages = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean().reset_index()
print(sales_averages)
```

```python
# Sales totals
sales_sums = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum().reset_index()
print(sales_sums)
```

## Sorting and Ranking Data

```python
# Ascending Values
print("\n ASCENDING: ")
ascended = df_sales.sort_values(by='Units_Sold', ascending = True)
print(ascended)
print("\n DESCENDING: ")
descended = df_sales.sort_values(by='Revenue', ascending = False)
print(descended)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False, method='dense').astype(int)
print(df_sales)
```

## Merging DataFrames

```python
# DF_PRODUCTS
product_data = {'Product': ['A', 'B', 'C'],
'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)
```

```python
# MERGER
df_merged = pd.merge(df_sales, df_products, on='Product')
print(df_merged)
```

```python
# Revenue by Category
category_revenue = df_merged.groupby('Category')['Revenue'].sum().reset_index()
print(category_revenue)
```
