# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('test_data.csv')
```

```python
df.describe()
```

## Filtering a DataFrame

```python
filtered_df = df[df['Age'] > 30]
filtered_df
```

```python
filtered_score = df[(df['Score'] > 50) & (df['Score'] < 100)]
filtered_score
```

## Modifying a DataFrame

```python
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['City'])
```

```python
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
print(df_cleaned)
```

```python
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
```

```python
df_cleaned = df_cleaned.dropna()
```

```python
df_cleaned
```

```python
df_cleaned.index = df_cleaned['ID'] # Index Labels
df_cleaned = df_cleaned.drop(columns=['ID']) # Removing column ID
```
