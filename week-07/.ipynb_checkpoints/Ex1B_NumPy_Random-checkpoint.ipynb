# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 20, size=2) # Two elements
print(r_1)
```

```python
r_2 = np.random.randint(1, 20, size = (1, 2)) # 1 row two columns
print(r_2)
```

```python
r_3 = np.random.uniform(0, 1, size = (2, 1))
print(r_3)
```

```python
r_4 = np.random.randint(20, 100, size = (4, 5))
print(r_4)
```

```python
r_5 = np.random.uniform(0, 1, size = (4, 5))
print(r_5)
```

## Generating Random Choices

```python
students = np.array([
    "Stan", "Kyle", "Cartman", "Kenny",
    "Butters", "Randy", "Wendy", "Token",
    "Jimmy", "Tweek"
])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size = (3, 4))
print(breakout_groups)

# This wouldn't be a very effective method of partitioning breakout groups because of duplicate assignments.
# Not possible to be in a breakout group with a clone of yourself...yet
```

```python
# Probabilities
prob = [0.10, 0.10, 0.30, 0.00, 0.10, 0.10, 0.10, 0.05, 0.10, 0.05]
# %30 chance of Cartman
# %0 chance of Kenny...you bastards!
```

```python
call_on_weighted = np.random.choice(students, size=5, replace=False, p=prob)
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
