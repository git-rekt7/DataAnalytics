# Lab 3. Employee Data
## This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
hr_list_temp = list(hr_list[1])
```

```python
hr_list_temp[2] = ('Mark Blick-Hawley')
```

```python
print(hr_list_temp)
```

```python
hr_list[1] = tuple(hr_list_temp)
```

```python
print(hr_list)
```

```python
hr_list_temp = list(hr_list[-1]) # tuple to list conversion
```

```python
hr_list_temp[-1] = ('60000')
```

```python
print(hr_list_temp) 
```

```python
hr_list[-1] = tuple(hr_list_temp)
```

```python
print(hr_list_temp) # successful update
```

```python
print(hr_list)
```

```python
print(hr_list[1], hr_list[-1]) # Pre Formatted print statement test
```

```python
print(f"{hr_list[1][0]}|{hr_list[1][1]}|{hr_list[1][2]}|{int(hr_list[1][3]):,}")
print(f"{hr_list[-1][0]}|{hr_list[-1][1]}|{hr_list[-1][2]}|{int(hr_list[-1][3]):,}")
```
