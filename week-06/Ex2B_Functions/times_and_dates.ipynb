## Working with datetime functions

# Today

```python
import datetime
```

```python
today = datetime.datetime.now()
```

```python
f = today.strftime("%A-%m-%d-%Y") 
```

```python
print(f"Today is {f}") # Returns day of week, month, day, year
```

# My Birthday

```python
Birthday = datetime.datetime(2004, 10, 23)
```

```python
print(Birthday)
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days = 90)
```

```python
print(f"90 days from today is {ninety_d}")
```

# Time for Dinner

```python
dinner_time = datetime.time()
```

```python
f = dinner_time.strftime("%H:%M:%S")
```

```python
print(f"Let's meet for dinner at {f}")
```
