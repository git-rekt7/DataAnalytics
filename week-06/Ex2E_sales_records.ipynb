This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument. 

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region 

```python
def region_total(sales_data, region):
    return sum(sale_amount for product, reg, sale_amount in sales_data if reg == region)
```

```python
print("\n EASTERN REGION TOTALS:")
print(region_total(sales_data, 'East')) # Sales total for eastern region
print("\n WESTERN REGION TOTALS:")
print(region_total(sales_data, 'West')) # Sales total for western region
print("\n SOUTHERN REGION TOTALS:")
print(region_total(sales_data, 'South')) # Sales total for southern region
```

## Lab 2. Count of sales by region

```python
def region_count(sales_data, region):
    return sum(1 for product, reg, sale_amount in sales_data if reg == region)
```

```python
print("\n EASTERN COUNT:")
print(region_count(sales_data, "East"))
print("\n WESTERN COUNT:")
print(region_count(sales_data, "West"))
print("\n SOUTHERN COUNT:")
print(region_count(sales_data, "South"))
print("\n NORHTERN COUNT:")
print(region_count(sales_data, "North"))
```

# Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    return [row for row in data if row[2] > threshold]
```

```python
results = (sales_above(sales_data, 500))
for row in results:
    print(row)
```

```python
def sales_below(data, threshold):
    return [row for row in data if row[2] < threshold]
```

```python
print("\n*BONUS*:")
results = sales_below(sales_data, 500)
for row in results:
    print(row)
```

# Lab 5. Top and bottom scale

```python
def top_sale(data):
    return max(data, key=lambda row: row[2])
```

```python
print(top_sale(sales_data)) # prints the top sale
```

```python
def bottom_sale(data):
    return min(data, key=lambda row: row[2])
```

```python
print(bottom_sale(sales_data))
```

The top sale was a laptop in the western region sold for 1300.00  
The lowest sale was a keyboard in the western region sold for 75.00

# Lab 6. Apply a discount

```python
def apply_discount(data, discount):
    return [
        (product, region, sale_amount * (1 - discount))
        for product, region, sale_amount in data
    ]
```

```python
sale_discount = apply_discount(sales_data, 0.15) # applies 15% discount
```

```python
# prints first 3 rows b4 discount
print("before discount:")
for row in sales_data[3]:
    print(row)
```

```python
# Prints first three rows after discount
print("After discount:")
for row in sale_discount[3]:
    print(row)
```

# Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set((row[1] for row in sales_data)))
```

```python
regions = unique_regions(sales_data)
print(regions)
```

This function could be useful in a large data set of say 1000 values. If I wanted to return all of the unique locations it would be too tedious to manually specify which regions to return. This function can be scaled up to any size.

# Lab 8. Region Summary

```python
def region_summary(data, region):
    # Filters specific records for a given region
    region_data = [row for row in data if row[1] == region]

    # Summary Stats
    total = region_total(data, region)
    count = region_count(data, region)
    average = total / count if count else 0 
    top = top_sale(region_data)

    # Formatted report print statement
    print(f"\n === Sales summary for {region} region ===")
    print(f"Total Sales: ${total:.2f}")
    print(f"Number of Sales: {count}")
    print(f"Average Sale Amount: ${average:.2f}")

    print("\nTop Sale:")
    print(f"Product: {top[0]}")
    print(f"Region: {top[1]}")
    print(f"Amount: ${top[2]:.2f}")
    
```

```python
region_summary(sales_data, "East")
```

# Lab 9. Full Summary

```python
for region in unique_regions(sales_data):
    region_summary(sales_data, region)
```
