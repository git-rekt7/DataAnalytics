## Working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f"{name}\nFavorite movie: {movie}\nFavorite music: {music}"
```

```python
print(favorite_things("Corey", "Donnie Darko", "Motörhead"))
```

```python
def why_im_here(purpose):
    return f"I joined the Data pathway to use information to {purpose}"
```

```python
print(why_im_here("make an impact"))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}"
```

```python
print(favorite_place("the practice room", "I can jam with my friends"))
```

```python
print(favorite_place("your mom's house", "we play Cards Against Humanity"))
```
