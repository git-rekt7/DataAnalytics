## Tax Functions

### The purpose of this lab is to determine tax rates according to various withholding codes

```python
# Social Security
def get_soc_sec_tax(gross_pay):
    tax_rate = 0.062
    return gross_pay
```

```python
# Medicare Tax
def get_medicare_tax(gross_pay):
    tax_rate = 0.0145
    return gross_pay
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else: # 4 or more
        tax_rate = 0.18

    return gross_pay * tax_rate
```

```python
#               Person One
print("\n PERSON ONE")
print("\n ~~~~~~~~~~~")
print(get_federal_tax(750, 0))
```

```python
#               Person Two
print("\n PERSON TWO")
print("\n ~~~~~~~~~~~")
print(get_federal_tax(1550, 2))
```
