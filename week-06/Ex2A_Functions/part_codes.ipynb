# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings - part codes

```python
supplier_info = "BlackRiver:666-M"
```

```python
def get_supplier_code(s):
    return s.split(":")[0] # Supplier name
```

```python
def get_product_num(s):
    return s.split(":")[1].split("-")[0] # Product Number
```

```python
def get_size(s):
    return s.split("-")[1] # Size 
```

```python
print(get_supplier_code(supplier_info)) # BlackRiver
```

```python
print(get_product_num(supplier_info)) # 666
```

```python
print(get_size(supplier_info)) # M
```

```python
print(f"For {supplier_info}, the supplier is {get_supplier_code(supplier_info)}. Part number {get_product_num(supplier_info)} is available in size {get_size(supplier_info)}.")
```
